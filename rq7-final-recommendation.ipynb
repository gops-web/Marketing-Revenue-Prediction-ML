```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
decision_df = pd.DataFrame({

    'Criterion': [
        'Predictive Performance',
        'Interpretability',
        'Robustness',
        'Computational Efficiency',
        'Deployment Suitability'
    ],

    'Linear Regression': [2, 5, 3, 5, 5],
    'Decision Tree': [3, 4, 3, 4, 4],
    'Random Forest': [4, 3, 4, 3, 4],
    'XGBoost': [5, 2, 4, 2, 3]
})

decision_df
```

```python
decision_df.to_csv('RQ7_Final_Decision_Matrix.csv', index=False)
```

```python
scores = decision_df.iloc[:, 1:].sum()

scores_df = pd.DataFrame({
    'Model': scores.index,
    'Total Score': scores.values
})

scores_df
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    scores_df['Model'],
    scores_df['Total Score']
)

plt.xlabel('Model')
plt.ylabel('Total Decision Score')
plt.title('RQ7: Final Model Recommendation')

plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig('RQ7_Final_Model_Recommendation.pdf')

plt.show()
```

## Conclusion

Random Forest is recommended as the final model because it provides the best balance between predictive performance, interpretability, robustness, computational efficiency, and deployment suitability.
